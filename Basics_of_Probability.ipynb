```python
import random
def pic_ball_experiment():
  balls=['blue','red','green']

  result=random.choice(balls)

  prob=balls.count('red')/len(balls)
  print(f"probability of picking a red ball is: {prob}")
  if result=="red":
    return 'Red ball was picked'
  else:
    return 'better luck next time'
res=pic_ball_experiment()
print(res)
```
